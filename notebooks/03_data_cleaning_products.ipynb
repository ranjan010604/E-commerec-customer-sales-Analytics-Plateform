1. importing os and pandas for file and data manipulation

```python
import os
import pandas as pd
```

2. Create new file directory to store ecleaned products data named'../data/cleaned'

```python
os.makedirs("../data/cleaned", exist_ok=True)
```

3. Loading Raw data from raw data folder

```python
products_clean = pd.read_csv("../data/raw/products_raw.csv")
```

4. Selecting string datatype folder for string manipulation and removing whitespace leading and trailing and stanadardize it

```python
text_columns = [
    "product_id",
    "product_name",
    "brand",
    "category",
    "sub_category",
]

for col in text_columns:
  products_clean[col] = products_clean[col].astype(str).str.strip()
```

5. Converting negative 'cost_price' value to absolute value to tackle the errors 

```python
products_clean["cost_price"] = products_clean["cost_price"].abs()
```

6. Fill Missing data of 'stock_quantity' with 0 and convert it into int datatypes

```python
products_clean["stock_quantity"] = (
    products_clean["stock_quantity"].fillna(0).astype(int)
)
```

7. Exporting cleaned data into a new folder named '../data/cleaned/products_cleaned.csv' and the exporting datatype is csv.

```python
output_path = os.path.join("../data/cleaned", "products_cleaned.csv")
products_clean.to_csv(output_path, index=False)
```
