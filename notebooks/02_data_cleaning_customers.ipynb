1. importing os and pandas for file and data mainpulation

```python
import os
import pandas as pd
```

2. Creating new directories named 'cleaned' to store cleaned data instead of cleaning raw data 

```python
os.makedirs("../data/cleaned", exist_ok=True)
```

3. Loading data from directories

```python
customers_clean =pd.read_csv("../data/raw/customers_raw.csv")
```

5. Selecting string columns for removing extra whitespace leading or trailing
6. Change 'signup_date' datatype from objects to datetime

```python
columns = [
    "gender",
    "city",
    "state",
    "customer_segment",
    "preferred_device"
]
for col in columns:
    customers_clean[col] = customers_clean[col].astype(str).str.strip()

    customers_clean["signup_date"] = pd.to_datetime(customers_clean["signup_date"], errors ="coerce")
```

7. calculating median age of the age column
8. Replacing missing age value with median age of the age column as per the buisnss rule
9. Replacing missing city of 'city' column with 'unknown' data

```python
median_age = customers_clean["age"].median()
customers_clean["age"] =  customers_clean["age"].fillna(median_age)
customers_clean["city"] =  customers_clean["city"].fillna("unknown")
```

10. export the cleaned data to external directories named '../data/cleaned/customers_cleaned.csv' folder
11. export folder type is csv

```python
output_path = os.path.join("../data/cleaned", "customers_cleaned.csv")
customers_clean.to_csv(output_path, index=False)
```
