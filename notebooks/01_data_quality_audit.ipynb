13. Inavlid Values Inspectiion

```python
import pandas as pd

invalid_records = []


inv_cost = (datasets['Products']['cost_price'] <= 0).sum()
invalid_records.append({
    'Dataset': 'Products',
    'Column': 'cost_price',
    'Issue': 'Negative cost price (<= 0)',
    'Count': inv_cost,
})


inv_qty = (datasets['Orders']['quantity'] <= 0).sum()
invalid_records.append({
    'Dataset': 'Orders',
    'Column': 'quantity',
    'Issue': 'Negative quantity (<= 0)',
    'Count': inv_qty,
})


inv_sold = (datasets['Inventory']['units_sold'] < 0).sum()
invalid_records.append({
    'Dataset': 'Inventory',
    'Column': 'units_sold',
    'Issue': 'Negative units sold (< 0)',
    'Count': inv_sold,
})


inv_gender_space = (
    datasets['Customers']['gender']
    .str.contains('^\s+|\s+$', regex=True)
    .sum()
)
invalid_records.append({
    'Dataset': 'Customers',
    'Column': 'gender',
    'Issue': 'Leading/trailing whitespace',
    'Count': inv_gender_space,
})



# Display summary dataframe
invalid_summary_df = pd.DataFrame(invalid_records)
display(invalid_summary_df)
```

creation of 2nd table

```python
import pandas as pd


missing_customer_id = datasets['Orders']['customer_id'].isnull().sum()


negative_quantity = (datasets['Orders']['quantity'] < 0).sum()


duplicate_order_id = datasets['Orders']['order_id'].duplicated().sum()


missing_rating = datasets['Orders']['customer_rating'].isnull().sum()


invalid_stock_products = datasets['Products']['stock_quantity'].isnull().sum()
invalid_stock_inventory = (
    datasets['Inventory']['closing_stock'].isnull().sum()
)


check_results = pd.DataFrame([
    {
        'Issue Checklist': 'Missing Customer ID',
        'Dataset': 'Orders',
        'Column': 'customer_id',
        'Count': missing_customer_id,
    },
    {
        'Issue Checklist': 'Negative Quantity',
        'Dataset': 'Orders',
        'Column': 'quantity',
        'Count': negative_quantity,
    },
    {
        'Issue Checklist': 'Duplicate Order ID',
        'Dataset': 'Orders',
        'Column': 'order_id',
        'Count': duplicate_order_id,
    },
    {
        'Issue Checklist': 'Missing Rating',
        'Dataset': 'Orders',
        'Column': 'customer_rating',
        'Count': missing_rating,
    },
    {
        'Issue Checklist': 'Invalid/Missing Stock',
        'Dataset': 'Products',
        'Column': 'stock_quantity',
        'Count': invalid_stock_products,
    },
    {
        'Issue Checklist': 'Invalid/Missing Stock',
        'Dataset': 'Inventory',
        'Column': 'closing_stock',
        'Count': invalid_stock_inventory,
    },
])

display(check_results)
```
